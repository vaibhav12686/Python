# **Basic Class and Object**

Create a car class with attributes like brand and model. Then create an instance of this class




````python
class Car:
    brand = None
    model = None

my_car = Car()          # object
print(my_car)
````

````python
class Car:
    def __init__(userbrand, usermodel):
        brand = userbrand
        model = usermodel

my_car = Car("Toyota", "Corolla")
print(my_car)

# python calls -->  Car.__init__(new_object, "Toyota", "Corolla")
# but there init only has two parameters - userbrand and usermodel


"""
my_car = Car("Toyota", "Corolla")

Python automatically creates a new Car object because you called the class like a function.

Conceptually:

new_object = Car.__new__(Car)
Car.__init__(new_object, "Toyota", "Corolla")
my_car = new_object

So the object was created, but Python handles the creation automatically.

__new__() creates the object, and __init__() initializes it.

"""
````

***Methods in Python***

---



A method is a function that belongs to a class/object.

Example:
```
class Car:
    def drive(self):
        print("Car is driving")
```

Here:
```
def drive(self):
```
is a method because it is a function defined inside a class.

You use it through an object:
```
my_car = Car()
my_car.drive()
```

Output:
`Car is driving`

````python
class Car:
    def __init__(self, brand, model):  # brand, model  -  parameter given by the user when creating an object  with class Car  - >  my_car = Car("Toyota", "Corolla")
        self.brand = brand    #  self.brand  -  talking about the variables/attributes in the class
        self.model = model

my_car = Car("Toyota", "Corolla")
print(my_car.brand)
print(my_car.model)


'''

self       .       brand       =       brand
 ↓                  ↓                   ↓
object          attribute           parameter

''';
````

***self***

A class is a blueprint.
When you call a class like `Car()`, Python creates an object.
self refers to that particular object.
self allows the methods inside the class to talk to/access the object's data.

Without self, a variable inside a method is usually just a local variable.

Example:
```
class Car:
    def __init__(self, brand):
        self.brand = brand

    def show(self):
        print(self.brand)

my_car = Car("Toyota")
my_car.show()
```

Think:

`self` → "the current object I'm working with"

So `self.brand` means:

"The brand belonging to this particular object."

````python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

my_car = Car("Toyota", "Corolla")
print(my_car.brand)
print(my_car.model)

my_new_car = Car("Tata", "Safari")
print(my_new_car.brand)
print(my_new_car.model)
````
