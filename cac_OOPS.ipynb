# **Inheritance**

Create an ElectricCar class that inherits from the Car class and has an additional attribute battery_size

````python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def full_name(self):
        return f"{self.brand} {self.model}"

class ElectricCar(Car):  #  -->  added the class name to inherit all from Class Car
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size


my_tesla = ElectricCar("Tesla", "Model-S", "85kWh")
print(my_tesla.brand)

print(my_tesla.full_name())
````

`super()` means “use the parent class.”
```
class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size
```
`ElectricCar` inherits from `Car`.

The parent `Car` already knows how to set:
```
self.brand = brand
self.model = model
```
So instead of writing that code again, we use:
```
super().__init__(brand, model)
```
This calls Car's __init__.

Then ElectricCar adds its own extra attribute:
```
self.battery_size = battery_size
```

# **Encapsulation**

Modify the Car class to encapsulate the brand attribute, making it private, and provide a getter method for it.

**Encapsulation** means keeping an object's data and the methods that work with that data together, while controlling how outside code accesses it.



```
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance
```

Here:

`self.__balance` is internal/private data.

Instead of allowing outside code to freely change it:

`account.__balance = -5000`

we provide controlled methods:

```
account.deposit(500)
account.get_balance()
```
_ vs __

`self.balance`

Public → freely accessible.

`self._balance`

Protected convention → "This is for internal use; don't touch it directly." Python doesn't enforce this.

`self.__balance`

Private/name-mangled → Python changes its internal name to discourage direct access and prevent accidental conflicts.

The key idea

Don't think:

`"Encapsulation = making variables private."`

Think:

`"Encapsulation = hiding/controling internal details and providing a proper way to interact with the object."`

Encapsulation = "Keep the inside protected; expose only what others need."

````python
class Car:
    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model

    def get_brand(self):
        return self.__brand + " !"

    def full_name(self):
        return f"{self.__brand} {self.model}"

class ElectricCar(Car):  #  -->  added the class name to inherit all from Class Car
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size


my_tesla = ElectricCar("Tesla", "Model-S", "85kWh")
#print(my_tesla.__brand)
print(my_tesla.get_brand())

#  but here we dont want the user to access the brand variable
# use double underscores  ->   it makes the variable private

# class can access the variable
# but the object created will not be able to access the variable

# if we want the objects to use the Class variable -->  we need to create methods for that object (get_brand)
````

# **Polymorphism**

Demonstrate polymorphism by defining a method fuel_type in both Car and ElectricCar classes, but with different behaviors

**Polymorphism** means: same method name, different behavior depending on the object.

Think:

One command → different objects → different responses.

Example:
```
class Car:
    def move(self):
        print("Car drives")

class Boat:
    def move(self):
        print("Boat sails")

class Plane:
    def move(self):
        print("Plane flies")
```
Now:

```
for vehicle in [Car(), Boat(), Plane()]:
    vehicle.move()
```

Output:

```
Car drives
Boat sails
Plane flies
```

Notice: we wrote the same command every time:

`vehicle.move()`

But Python gives different behavior because vehicle is a different object.

The key distinction
Polymorphism is not:

"Different methods have different names."

It is:

"Different objects respond to the same method/interface in their own way."

Think of move() as a common instruction:
```
move()
 ↓
Car   → drive
Boat  → sail
Plane → fly
```
This is especially useful with inheritance:
```
class Animal:
    def speak(self):
        pass

class Dog(Animal):
    def speak(self):
        print("Woof")

class Cat(Animal):
    def speak(self):
        print("Meow")
```
Both have speak(), but each implements it differently.

One sentence to never forget
Polymorphism = same interface/method, different implementation/behavior.

````python
class Car:
    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model

    def get_brand(self):
        return self.__brand + " !"

    def full_name(self):
        return f"{self.__brand} {self.model}"

    def fuel_type(self):
        return "Petrol or Diesel"


class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

    def fuel_type(self):
        return "Electric Charge"


my_tesla = ElectricCar("Tesla", "Model-S", "85kWh")
print(f"Tesla model - {my_tesla.fuel_type()}")

safari = Car("Tata", "Safari")
print(safari.fuel_type())
````

# **Class Variables**

Add a class variable to Car that keeps track of the number of cars created

add a class variable such as `number_of_cars = 0`, then increase it every time `__init__` runs.

````python
class Car:

    total_cars = 0

    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model

        Car.total_cars += 1

    def get_brand(self):
        return self.__brand + " !"

    def full_name(self):
        return f"{self.__brand} {self.model}"

    def fuel_type(self):
        return "Petrol or Diesel"


class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

    def fuel_type(self):
        return "Electric Charge"


# CAR 1
my_tesla = ElectricCar("Tesla", "Model-S", "85kWh")
print(f"Tesla model - {my_tesla.fuel_type()}")

# CAR 2
safari = Car("Tata", "Safari")
print(safari.fuel_type())

print(f"total number of cars created = {Car.total_cars}")
````

# **Static Method**

Add a static method to the Car class that returns a general description of a car

````python
class Car:

    total_cars = 0

    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model

        Car.total_cars += 1

    def get_brand(self):
        return self.__brand + " !"

    def full_name(self):
        return f"{self.__brand} {self.model}"

    def fuel_type(self):
        return "Petrol or Diesel"

    @staticmethod
    def general_description():
        return "Cars are a means of transport"


class ElectricCar(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

    def fuel_type(self):
        return "Electric Charge"


print(Car.general_description())

# does not require --  self
````

A static method is a method that belongs to the class but does NOT need an object (self) to work.

```
@staticmethod
def general_description():
    return "Cars are means of transport"
```
Notice there is no self.

Why? Because this method doesn't need information about a particular car.

It doesn't need:
```
self.__brand
self.model
self.battery_size
```

It simply gives general information about cars.

So you can call it directly using the class:

`print(Car.general_description())`

Compare the three
```
def full_name(self):
```
→ needs a specific object
```
def get_brand(self):
```
→ needs a specific object
```
@staticmethod
def general_description():
```
→ needs neither the object nor class

Think:

Instance method
self → "this particular car"

Static method
        → "general functionality related to cars"
