# **DECORATORS**

Python Decorators as a Toll Booth 🚗🛣️

Imagine a toll booth on a highway.

Your function is the car traveling on the highway.

A decorator is the toll booth placed around that road.

The car's actual destination doesn't change, but the toll booth can do something before or after the car passes.

Without a decorator
```
def drive():
    print("Driving to destination")
drive()
```

The function simply runs.

```
With a decorator
def toll_booth(func):
    def wrapper():
        print("Collect toll")
        func()
        print("Open barrier")

    return wrapper
```

Now apply it:
```
@toll_booth
def drive():
    print("Driving to destination")
```

When Python sees:
```
@toll_booth
def drive():
```

it essentially does:

`drive = toll_booth(drive)`

This is the most important thing to understand.

The decorator receives the original drive function and returns a new function (wrapper).

So when you later do:

`drive()`

you're actually calling the wrapper:

```
drive()
   ↓
wrapper()
   ↓
"Collect toll"
   ↓
original drive()
   ↓
"Driving to destination"
   ↓
"Open barrier"
```

Why decorators exist
They let you add behavior to functions without modifying the function's original code.

For example, decorators are commonly used for:
```
Logging

Authentication/authorization

Timing functions

Caching

Validation

Access control
```

The complete mental model
Think:
```
Function = car
Decorator = toll booth around the car's route
Wrapper = the booth's process
@decorator = attach the toll booth to the function
```
And remember:

A decorator takes a function → modifies/wraps its behavior → returns a function.

That is the core concept.

# **for understanding**

1. The decorator does NOT modify the original function
Think of it like this:

Original function = Car
Decorator        = Toll booth
Wrapper           = New route through the toll booth

The original car still exists. The decorator wraps it with extra behavior.
```
def toll_booth(func):
    def wrapper():
        print("Collect toll")
        func()
        print("Open barrier")

    return wrapper
```

func is the original function.

wrapper is the new function that controls when/how the original function runs.

2. `@toll_booth` is just shorter syntax
This:
```
@toll_booth
def drive():
    print("Driving")
```

is essentially:
```
def drive():
    print("Driving")

drive = toll_booth(drive)
```

So after decoration, the name drive now refers to the wrapper.

That's why calling:

`drive()`

goes through the toll booth first.

3. The decorator can control the original function
The wrapper can decide:
```
print("Before")
func()
print("After")
```

Or even:

```
print("Checking permission")

if allowed:
    func()
```

So a decorator isn't merely "extra code before/after."

It can control whether, when, and how the original function executes.

4. Real decorators usually handle arguments
Your first example has no arguments:

`def wrapper():`

But functions often do:
```
def drive(destination):
    print(destination)
```

So decorators commonly use:
```
def toll_booth(func):
    def wrapper(*args, **kwargs):
        print("Collect toll")
        result = func(*args, **kwargs)
        print("Open barrier")
        return result

    return wrapper
```

*args and **kwargs basically mean:

"Whatever arguments the original function accepts, pass them through."
```
Final mental model
                 DECORATOR
                     ↓
Call function → [ TOLL BOOTH ] → Original function
                     ↑
              extra behavior
```
Decorator = a function that takes another function, wraps it with additional behavior, and returns the wrapped function.

# **Problem 1: Timing Function Execution**

write a decorator that measures the time a function takes to execute

````python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} ran in {end-start} time")
        return result
    return wrapper

@timer
def example_func(n):
    time.sleep(n)

example_func(2)
````

# **Problem 2: Debugging Function Calls**

Create a decorator to print the function name and the values of its arguments every time the function is called

````python
def debug(func):
    def wrapper(*args, **kwargs):
        args_value = ', '.join(str(arg) for arg in args)
        kwargs_value = ', '.join(f"{k}={v}" for k,v in kwargs.items())
        print(f"calling: {func.__name__} with args {args_value} and kwargs {kwargs_value}")
        return func(*args, **kwargs)

    return wrapper

@debug
def hello():
    print("hello")

@debug
def greet(name, greeting="hello"):
    print(f"{greeting}, {name}")

hello()
greet("chai", greeting = "Hii")
````

# **Problem 3: Cache Return Values**

Implement a decorator that caches the return values of a function, so that when it's called with the same arguements, the cached value if returned instead of re-executing the function

````python
import time

def cache(func):
    cache_value = {}
    print(cache_value)
    def wrapper(*args):
        if args in cache_value:
            return cache_value[args]

        result = func(*args)
        cache_value[args] = result
        return result
    return wrapper

@cache
def long_running_func(a,b):
    time.sleep(4)
    return a + b

print(long_running_func(2, 3))
print(long_running_func(2, 3))
````
