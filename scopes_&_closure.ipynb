**Scope** = the region where a variable is accessible. Python follows the LEGB rule:

L — **Local**: Variables inside the current function.

E — **Enclosing**: Variables in an outer function when functions are nested.

G — **Global**: Variables defined at module level.

B — **Built-in**: Python's built-ins, e.g. len, print.

```python
x = 10

def outer():
    y = 20
    def inner():
        print(x, y)   # Global x, Enclosing y
    inner()
```

```python
x = 10

def f():
    global x
    x = 20
```

```python
def outer():
    x = 10
    def inner():
        nonlocal x
        x += 1
    inner()
```

**Closures**:
A closure is a function that remembers variables from its enclosing scope even after the outer function finishes.

Closures allow you to create functions with remembered data.



```python
def multiplier(n):
    def multiply(x):
        return x * n
    return multiply

double = multiplier(2)
print(double(5))  # 10
```
