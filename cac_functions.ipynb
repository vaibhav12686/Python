```python
# basic function syntax

# write a function to calculate and return the square of a number

def square(num):
  print(num ** 2)

result = square(8)
print(result)
```

```python
def square(num):
  return num ** 2

result = square(8)
print(result)
```

```python
# function with multiple parameters

# create a function that takes two numbers as parameters and returns their sum

def sum_of_two(num1, num2):
  return num1 + num2

result = sum_of_two(92,88)
print(result)
```

```python
# polymorphism in functions

# write a function "multiply" that multiplies two numbers, but can also accept and multiply strings

def multiply(p1, p2):
  return p1 * p2

print(multiply(6, 12))
print(multiply("d", 5))
```

```python
# function returning multiple values

# create a function that returns both the area and circumference of a circle given its radius

import math

def circle(radius):
  area =  math.pi * radius ** 2
  circumference = 2 * math.pi * radius
  return area, circumference

a, c = circle(5)
print(f"Area of circle is {a:.2f},\nCircumference of circle is {c:.2f}")
```

```python
# default parameter value

# write a function that greets a user. If no name is provided, it should greet with a default name

def greet(name = 'User'):
  return "Hello, " + name + "!"

print(greet("Vaibhav"))
print(greet())
```

```python
# lambda function - A lambda function is a small, nameless function used in programming to run a single expression quickly without a formal definition.

# create a lambda function to compute the cube of a number

cube = lambda x: x ** 3

print(cube(3))
```

```python
# function with *args

# write a function that takes variable number of arguments and returns their sum

def sum_all(*args):
  return sum(args)

print(sum_all(1,2,3,4,5))
print(sum_all(1,2,3,4,5,6,7,8,9))
```

```python
# print(*args) - will return the numbers

# print (args) - will return a tuple of numbers

def sum_all(*args):
  print(args)
  for i in args:
    print(i * 2)

  return sum(args)

print(sum_all(1,2,3,4))
```

```python
# function with **kwargs

# create a function that accepts any number of keywords arguments and prints them in the format key : value

def print_kwargs(name, power):
  print("Name: ", name, " Power: ", power)

print_kwargs(name = "Superman", power = "lazer")
```

```python
def print_kwargs(**kwargs):
  for key, value in kwargs.items():
    print(f"{key} : {value}")

print_kwargs(name = "Superman", power = "Lazer")
print_kwargs(name = "Superman")
print_kwargs(name = "Superman", power = "lazer", enemy = "Lex Luthor")
```

```python
# generator function with yield

# write a generator function that yeilds even numbers up to a specified limit

def even_generator(limit):
  list1 = []
  for i in range(2, limit + 1, 2):
    list1.append(i)
  return list1

print(even_generator(10))
```

```python
def even_generator(limit):
  for i in range(2, limit + 1, 2):
    yield i

for num in even_generator(10):
  print(num)
```

```python
# recursive function

# create a recursive function to calculate the factorial of a number

def factorial(n):
  if n == 0:
    return 1

  else:
    return n * factorial(n-1)

print(factorial(5))
```
